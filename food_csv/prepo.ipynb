```python
food_2 = food_2.drop(columns=["Unnamed: 0"])
```

```python
food = pd.concat([food1, food_2], ignore_index=True)
```

```python
food["food_id"] = range(1, len(food) + 1)
```

```python
food=food[['food_id','name','diet','course','flavor_profile','state','img_url']]
```

```python
food.head()
```

```python
food['tags'] = food['flavor_profile'] + " " + food['course'] + " " + food['state']+ " " + food['diet']
```

```python
new_df = food[['food_id','name','tags','img_url']]
```

```python
new_df.head(2)
```

VECTORIZATION

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cv = CountVectorizer(max_features=5000, stop_words='english')
vectors = cv.fit_transform(food["tags"]).toarray()
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
def recommend(food):

    food_idx = new_df[new_df['name'] == food].index[0]

    dist = similarity[food_idx]

    food_list = sorted(
        list(enumerate(dist)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    for i in food_list:
        print(new_df.iloc[i[0]]['name'])
```

```python
recommend('Dahi vada')
```
